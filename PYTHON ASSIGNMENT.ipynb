### GENRE ANALYSIS

#### Total global sales per genre. 

```python
GenreTotal = df.groupby('Genre')['Global_Sales'].sum()
GenreTotal.nlargest(12)
```

#### Average Sales per genre

```python
df.groupby('Genre')['Global_Sales'].mean()
```

#### Count of sales per genre

```python
GenreCount = df.groupby('Genre')['Name'].count()
GenreCount
```

## VISUALIZATIONS

```python
#Number of games per genre

sns.histplot(y = df.Genre,color = 'skyblue',edgecolor = 'black')

plt.show
```

```python
## Global sales per genre
plt.figure(figsize= (6,4))
GenreTotal.plot(kind= 'bar', color = 'green')
width = 0.4


plt.title('Total global sales per genre')
plt.xlabel('Genres')
plt.ylabel('Total global sales')

plt.show()
```

```python
plt.figure(figsize= (12,10))
x = np.arange(12)
width = 0.4

#Plotting a grouped bar chart
plt.bar(x-0.2,GenreCount,width, color = 'skyblue')
plt.bar(x+0.2,GenreTotal,width, color = 'pink')
plt.xticks(x, ['Sports','Platform','Racing','Role-Playing','Puzzle','Misc','Shooter'
              ,'Simulation','Action','Fighting','Adventure','Strategy'])

#Adding values on top of the bars


plt.title('Total global sales per genre')
plt.xlabel('Genres')
plt.ylabel('Total global sales')
plt.legend(['Count','G.Sales'])
plt.show()
```

### SECTION 3: TIME TREND ANALYSIS.

#### SALES OVER TIME

```python
YSales = df.groupby('Year')['Global_Sales'].sum()
YSales
```

```python
NSales = df.groupby('Year')['NA_Sales'].sum()
NSales
```

```python
ESales = df.groupby('Year')['EU_Sales'].sum()
ESales
```

```python
JSales = df.groupby('Year')['JP_Sales'].sum()
JSales
```

```python
YSales.plot(kind = 'line', linestyle='-',color = 'red')
NSales.plot(kind = 'line', linestyle='-',color = 'green')
ESales.plot(kind = 'line', linestyle='-',color = 'blue')
JSales.plot(kind = 'line', linestyle='-',color = 'yellow')


plt.xlabel("Year")
plt.ylabel("Global Sales")
plt.title("Global Sales Trend 1985-Present")
plt.legend()
plt.grid(True)
plt.show()

```

### STEP 4: PUBLISHER STRATEGY ANALYSIS.

#### TOP 10 PUBLISHERS BY GLOBAL SALES.

```python
TopPublish = df.groupby('Publisher')['Global_Sales'].sum()
TopPublish.nlargest(10)
```

```python
PublishCount = df.groupby('Publisher')['Name'].count()
PublishMean=TopPublish/PublishCount
PublishMean.nlargest(10)
```

```python
PublishC = PublishCount.nlargest(10)
PublishC
```

```python
##Global sales per publisher
plt.figure(figsize= (6,4))
PublishC.plot(kind= 'bar', color = 'pink')
x = np.arange(10)


#Plotting a grouped bar chart

plt.xticks(x, ['EA', 'Acti',' NBG', 'Ubi', 'KDE', 'THQ', 'Nin', 'SCE', 'Sega',' TTI'])
plt.title('Total global sales per publisher')
plt.xlabel('Publishers')
plt.ylabel('Total global sales')

plt.show()
```
